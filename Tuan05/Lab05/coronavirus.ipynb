```python
import numpy as np
import pandas as pd
from scipy import stats
```

# 1. Load the .csv into a dataframe using read_csv 
(Tải file .csv vào một DataFrame bằng hàm read_csv)
+ Trong file data thì là code thay vì iso_code, country thay vì location trong bài mẫu
+ sửa tên theo file data
+ các file data sẽ được dẫn bằng link vì quá nặng

```python
covid_data = pd.read_csv("https://catalog.ourworldindata.org/garden/covid/latest/compact/compact.csv")
covid_data = covid_data[['code','continent','country','date','total_cases','new_cases']]
```

# 2. Take a quick look at the data
(Lọc chỉ lấy các cột cần thiết)
+ Xem nhanh 5 dòng đầu tiên của dữ liệu
+ Kiểm tra kiểu dữ liệu của các cột
+ Kiểm tra kích thước (số dòng, số cột) của tập dữ liệu

```python
covid_data.head(5)
covid_data.dtypes
covid_data.shape
```

# 3. Get the mean of the data
(Tính giá trị trung bình (Mean) của số ca mắc mới)

```python
data_mean = np.mean(covid_data["new_cases"])
```

# 4. Get the median of the data
(Tính giá trị trung vị (Median) của số ca mắc mới)

```python
data_median = np.median(covid_data["new_cases"])
```

# 5. Get the mode of the data
(Tìm yếu vị / giá trị xuất hiện nhiều nhất (Mode))

```python
data_mode = stats.mode(covid_data["new_cases"])
```

# 6. Obtain the variance of the data
(Tính phương sai (Variance) của dữ liệu)

```python
data_variance = np.var(covid_data["new_cases"])
```

# 7. Obtain the standard deviation of the data
(Tính độ lệch chuẩn (Standard Deviation))

```python
data_sd = np.std(covid_data["new_cases"])
```

# 8. Compute the maximum and minimum values of the data
(Tìm giá trị lớn nhất (Max) và nhỏ nhất (Min))

```python
data_max = np.max(covid_data["new_cases"])
data_min = np.min(covid_data["new_cases"])
```

# 9. Obtain the 60th percentile of the data
(Tính bách phân vị thứ 60 (60th Percentile))

```python
data_percentile = np.percentile(covid_data["new_cases"],60)
```

# 10. Obtain the quartiles of the data
(Tính tứ phân vị (ở đây là Q3 - Percentile thứ 75))

```python
data_quartile = np.quantile(covid_data["new_cases"],0.75)
```

# 11. Get the IQR of the data
(Tính khoảng tứ phân vị (IQR = Q3 - Q1))

```python
data_IQR = stats.iqr(covid_data["new_cases"])
```

```python
# In tất cả kết quả thống kê ra console (Đã fix lỗi Mode)
print(f"Mean: {data_mean}")
print(f"Median: {data_median}")
print(f"Mode: {np.ravel(data_mode.mode)[0]}")
print(f"Variance: {data_variance}")
print(f"Standard Deviation: {data_sd}")
print(f"Min: {data_min} | Max: {data_max}")
print(f"Range: {data_max - data_min}")
print(f"60th Percentile: {data_percentile}")
print(f"Q3 (75th Percentile): {data_quartile}")
print(f"IQR: {data_IQR}")
```
