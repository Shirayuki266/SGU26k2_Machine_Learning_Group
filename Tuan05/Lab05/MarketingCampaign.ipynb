## 2. Loại bỏ dữ liệu trùng lặp
+ axis = 0 là dòng, axis = 1 là cột.
+ 2 lệnh đầu là bắt buộc, tất yếu sài để kiểm tra dữ liệu và loại bỏ các dòng bị trùng lập.
+ 2 lệnh cuối là xóa dòng, cột. chúng xuất hiện trong bài với mục đích là dạy ta cách xóa dòng, cột.
+ vì nếu xóa thật thì phải **gán đè biến** hoặc thêm tham số **inplace=true** (xóa thật ở đây là xóa ở dữ liệu gốc)
+ xóa dòng thì dùng để xóa các dữ liệu rác, lỗi,... vì dữ liệu thực tế rất bẩn
+ xóa cột thì dùng khi cột đó không có giá trị phân tích, dư thừa. xóa để nó sạch, trống, dễ nhìn.

### 2.1. Loại bỏ các dòng bị trùng lặp hoàn toàn trên tất cả các cột trong tập dữ liệu:

## 3. Thay thế dữ liệu và thay đổi định dạng của dữ liệu
+ Chuyển đổi dữ liệu định dạng số (0, 1, 2) của cột Teenhome sang dạng chữ dễ đọc, dễ hiểu hơn.
+ làm Khi số lượng nhóm giá trị ít (ví dụ: 0, 1, 2) và ta muốn gom nhóm lại thành các trạng thái phân loại cụ thể (kém/yếu/trungbinh/tốt,...)
+ còn điền khuyết giá trị phải làm chứ không là làm lệch dữ liệu, hệ thống báo lỗi.
+ đổi float -> int Khi phát hiện kiểu dữ liệu bị Pandas nhận diện sai lúc đọc file

### 3.1. Thay thế các giá trị trong cột Teenhome thành 'has teen' (có con tuổi teen) và 'has no teen' (không có)


```python
marketing_data['Teenhome_replaced'] = marketing_data['Teenhome'].replace([0,1,2],['has no teen','has teen','has teen'])
```

### 3.2. Điền các giá trị bị khuyết (NaN/NAs) trong cột Income bằng số 0


```python
marketing_data['Income'] = marketing_data['Income'].fillna(0)
```

### 3.3. Thay đổi kiểu dữ liệu của cột Income từ số thực (float) sang số nguyên (int)


```python
marketing_data['Income_changed'] = marketing_data['Income'].astype(int)
```
